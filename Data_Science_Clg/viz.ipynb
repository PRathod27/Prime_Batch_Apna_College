```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.random.randint(1,100,50)
x = np.arange(1,51)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0,0].plot(x,data,color="white",marker="o")
axes[0,0].set_title('Line Chart')
axes[0,0].set_xlabel('Index')
axes[0,0].set_ylabel('Value')
axes[0,0].grid(True)


axes[0,1].scatter(x,data,color="white")
axes[0,1].set_title('Scatter Plot')
axes[0,1].set_xlabel('Index')
axes[0,1].set_ylabel('Value')
axes[0,1].grid(True)

axes[1,0].hist(data, bins=10, color="green",edgecolor="black")
axes[1,0].set_title('Histogram')
axes[1,0].set_xlabel('Index')
axes[1,0].set_ylabel('Value')
axes[1,0].grid(True)

axes[1,1].boxplot(data, patch_artist = True, boxprops={"facecolor": "lightblue"})
axes[1,1].set_title('BoxPlot')
axes[1,1].set_ylabel('Value')
axes[1,1].grid(True)

plt.tight_layout()
plt.show()
```
