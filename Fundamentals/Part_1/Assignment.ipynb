```python
# swap values
value1 = input("Enter first value: ")
value2 = input("Enter second value: ")
temp = value1
print(f"Before swapping value1 : {value1} and value2 : {value2}")
value1 = value2
value2 = temp
print(f"After swapping value1 : {value1} and value2 : {value2}")
```

```python
celsius = input("Enter temperature in Celsius: ")
celsius = float(celsius)
fahrenheit = (celsius * (9/5)) + 32
print(f"{celsius} degree Celsius is equal to {fahrenheit} degree Fahrenheit.")
```

```python
# area
radius = int(input("Enter radius : "))
area = 3.14 * radius ** 2
print(f"Area of circle with radius {radius} is {area}")
```

```python
principal = input("Enter principal amount: ")
rate = input("Enter rate of interest: ")
time = input("Enter time in years: ")
principal = float(principal)
rate = float(rate)
time = float(time)
print(f"{(principal * rate * time) / 100}")
```
