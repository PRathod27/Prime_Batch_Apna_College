```python
salary = int(input("Enter the salary: "))

if salary<30000:
    salary = salary * (5/100)
elif (salary>=30000 and salary <= 70000):
    salary = salary * (15/100)
else:
    salary = salary * (25/100)

print(salary)
```

```python
def eve_num(a,b):

    start = min(a,b)
    end = max(a,b)

    for i in range(start,end+1):
        if(i % 2 == 0):
            print(i)


eve_num(5,10)
```
