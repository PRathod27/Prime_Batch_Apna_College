```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
insurance_data = pd.read_csv('insurance.csv')
insurance_data.head()
```

```python
sns.set_style("white")
sns.scatterplot(x=insurance_data["bmi"], y=insurance_data["charges"], color = "blue", hue = insurance_data["smoker"], palette = "Set1")
```

```python
X = insurance_data.drop(columns=['charges',"region"])
Y = insurance_data["charges"]

X["sex"] = X["sex"].map({"female":1,"male":0})
X["smoker"] = X["smoker"].map({"yes":1,"no":0})
```

```python
X.head()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score
print("r-squared",r2_score(Y_test,y_pred))
```

```python
r2 = r2_score(Y_test,y_pred)
```

```python
n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1-((1-r2)*(n-1)/(n-p-1))   #   ====> formula 
print(f"Adjusted r^2 value is: {adjusted_r2}")
```
